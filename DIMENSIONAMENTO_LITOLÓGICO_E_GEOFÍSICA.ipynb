```python
# AUTOR: MARCELO DE OLIVEIRA SANTOS JUNIOR----- MOSJ
# Geologia: Perfilagem Geofísica e Classificação Litológica 3D/4D


import numpy as np
import matplotlib.pyplot as plt

# 1. Simulação de perfil de poço (2000m a 2300m)
rng = np.random.default_rng(7)
depth = np.linspace(2000, 2300, 300)

# Folhelho base + zona de arenito reservatório (2120m - 2180m) + calcário (2230m - 2260m)
gr = rng.normal(95, 8, 300)
rhob = rng.normal(2.55, 0.04, 300)
nphi = rng.normal(0.28, 0.02, 300)
sw = np.ones(300) * 0.95  # Saturação de água

# Reservatório de Arenito com Hidrocarboneto
res_mask = (depth >= 2120) & (depth <= 2180)
gr[res_mask] = rng.normal(32, 5, res_mask.sum())
rhob[res_mask] = rng.normal(2.22, 0.03, res_mask.sum())
nphi[res_mask] = rng.normal(0.18, 0.02, res_mask.sum())
sw[res_mask] = rng.uniform(0.20, 0.35, res_mask.sum())  # Óleo/Gás presente

# 2. Visualização 2D (Track de Perfil) + 3D/4D (Crossplot Petrofísico)
fig = plt.figure(figsize=(13, 6))

ax1 = fig.add_subplot(1, 2, 1)
ax1.plot(gr, depth, color="darkgreen", lw=1.2)
ax1.fill_betweenx(depth, gr, 60, where=(gr < 60), color="gold", alpha=0.6, label="Arenito (Potencial Reservatório)")
ax1.invert_yaxis()
ax1.set_title("Perfil de Raios Gama (2D)")
ax1.set_xlabel("Gamma Ray (API)")
ax1.set_ylabel("Profundidade (m)")
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2 = fig.add_subplot(1, 2, 2, projection="3d")
sc = ax2.scatter(gr, rhob, nphi, c=(1 - sw) * 100, cmap="plasma", s=25)
cbar = plt.colorbar(sc, ax=ax2, pad=0.1, shrink=0.7)
cbar.set_label("Saturação de Hidrocarbonetos - 4D (%)")
ax2.set_title("Crossplot Petrofísico 3D + Saturação 4D")
ax2.set_xlabel("GR (API)")
ax2.set_ylabel("Densidade RHOB (g/cm³)")
ax2.set_zlabel("Porosidade NPHI (v/v)")

plt.tight_layout()
plt.show()
```
